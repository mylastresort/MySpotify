# Bonus -- Three Spotify-Inspired Features

Reproducing 3 recommendation blocks found on real music streaming services,
implemented on top of the same 4 dataset files used in the mandatory part.

| # | Feature | Spotify equivalent | Data used |
|---|---|---|---|
| 1 | Because you listened to [Artist] | "Because you listened to..." row | triplets + tracks |
| 2 | Your [Genre] Mix | "Daily Mix" | triplets + genres + tracks |
| 3 | Fans also like | "Fans also like" on artist page | triplets + tracks |

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

from src.data.loader import MySpotifyRecommender
from src.models.spotify_features import (
    because_you_listened_to,
    build_artist_user_matrix,
    fans_also_like,
    your_genre_mix,
)
```

```python
rs = MySpotifyRecommender.from_files()
```

```python
sample_user = (
    rs.triplets.groupby("user_id")["play_count"].sum()
    .sort_values(ascending=False)
    .index[0]
)
```

### Bonus 1 -- Because you listened to [Artist]

```python
recs1, seed_artist = because_you_listened_to(rs, sample_user)
print(f"Because you listened to: {seed_artist}\n")
display(recs1)
```

### Bonus 2 -- Your [Genre] Mix

```python
recs2, top_genre = your_genre_mix(rs, sample_user)
print(f"Your {top_genre} Mix\n")
display(recs2)
```

### Bonus 3 -- Fans also like

```python
artist_matrix, artist_idx, idx_artist = build_artist_user_matrix(rs)
print(f"Artist-user matrix: {artist_matrix.shape}")
```

```python
recs3 = fans_also_like(rs, seed_artist, artist_matrix, artist_idx, idx_artist)
print(f"Fans of {seed_artist} also like:\n")
display(recs3)
```
