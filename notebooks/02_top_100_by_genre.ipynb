# 02 -- Top 100 Most-Played Songs for a Given Genre

Returns the **100 globally most-played songs for a given majority genre**, sorted by total play count descending.

### Algorithm
1. Deduplicate `tracks` by `song_id` (keep the first `track_id`) so a song whose genre tag sits on a non-canonical track row is not double-counted.
2. Join with `genres` (via `track_id`) using only `majority_genre`.
3. Join with `triplets` (via `song_id`) and aggregate `play_count` (sum) per `song_id`.
4. Filter to the requested genre, keep the top 100 by total play count.

### Output columns
| Column | Description |
|--------|-------------|
| `rank` | 1-100 index, by total play count (descending) |
| `artist` | Artist name |
| `title` | Track title |
| `play_count` | Total plays across all users |
