# 01 -- Top 250 Most-Played Songs

Returns the **250 globally most-played songs** across all users, ranked by total play count.

### Algorithm
1. Aggregate `play_count` (sum) per `song_id` from the triplets.
2. Rank songs by total play count descending and keep the top 250.
3. Enrich with artist / title from `tracks`.

### Output columns
| Column | Description |
|--------|-------------|
| `rank` | 1-250, by total play count |
| `artist` | Artist name |
| `title` | Track title |
| `play_count` | Total plays across all users |
